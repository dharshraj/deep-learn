```python
# =====================================================================
# Question 17: Predict student performance, demonstrate overfitting & fix
# =====================================================================
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.regularizers import l2
from sklearn.model_selection import train_test_split

print("=== QUESTION 17: Overfitting Demonstration and Mitigation ===")
# Generate synthetic exam performance data
np.random.seed(42)
X = np.random.rand(100, 10)
y = (X[:, 0] * 3 + X[:, 1] * 2 + np.random.randn(100) * 0.1) > 2.5
y = y.astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Overfitted Model (No regularization, very complex)
model_overfit = Sequential([
    Input(shape=(10,)),
    Dense(512, activation='relu'),
    Dense(256, activation='relu'),
    Dense(1, activation='sigmoid')
])
model_overfit.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history_overfit = model_overfit.fit(X_train, y_train, epochs=150, validation_data=(X_test, y_test), verbose=0)

# Mitigated Model (Using Dropout & L2 Weight Regularization)
model_regularized = Sequential([
    Input(shape=(10,)),
    Dense(64, activation='relu', kernel_regularizer=l2(0.01)),
    Dropout(0.5),
    Dense(32, activation='relu', kernel_regularizer=l2(0.01)),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])
model_regularized.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history_regularized = model_regularized.fit(X_train, y_train, epochs=150, validation_data=(X_test, y_test), verbose=0)

print(f"Overfitted Model -> Train Acc: {history_overfit.history['accuracy'][-1]:.4f}, Test Acc: {history_overfit.history['val_accuracy'][-1]:.4f}")
print(f"Regularized Model -> Train Acc: {history_regularized.history['accuracy'][-1]:.4f}, Test Acc: {history_regularized.history['val_accuracy'][-1]:.4f}")
```

```python
# =====================================================================
# Question 18: Polynomial regression bias-variance trade-off
# =====================================================================
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

print("=== QUESTION 18: Bias-Variance Trade-off in Polynomial Regression ===")
np.random.seed(42)
X = np.sort(5 * np.random.rand(80, 1), axis=0)
y = np.sin(X).ravel() + np.random.normal(0, 0.2, X.shape[0])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

for degree in [1, 3, 15]:
    poly = PolynomialFeatures(degree=degree)
    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)

    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    train_err = mean_squared_error(y_train, model.predict(X_train_poly))
    test_err = mean_squared_error(y_test, model.predict(X_test_poly))
    print(f"Degree {degree:2d} | Train MSE: {train_err:.4f} | Test MSE: {test_err:.4f}")
```

```python
# =====================================================================
# Question 19: L1 and L2 regularization effects on model coefficients
# =====================================================================
import numpy as np
from sklearn.linear_model import Lasso, Ridge

print("=== QUESTION 19: L1 (Lasso) and L2 (Ridge) Regularization ===")
np.random.seed(42)
X = np.random.randn(100, 10)
y = X[:, 0] * 5 + X[:, 1] * -3 + np.random.randn(100) * 0.5

# L1 Lasso
lasso = Lasso(alpha=0.5)
lasso.fit(X, y)

# L2 Ridge
ridge = Ridge(alpha=10.0)
ridge.fit(X, y)

print("Lasso Coefficients (L1):", np.round(lasso.coef_, 3))
print("Ridge Coefficients (L2):", np.round(ridge.coef_, 3))
```

```python
# =====================================================================
# Question 20: Learning rates effect on convergence
# =====================================================================
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

print("=== QUESTION 20: Impact of Learning Rates on Loss Convergence ===")
learning_rates = [0.1, 0.01, 0.0001]
for lr in learning_rates:
    model = Sequential([Input(shape=(10,)), Dense(1)])
    model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=lr), loss='mse')
    history = model.fit(X, y, epochs=10, verbose=0)
    print(f"LR: {lr:<6} | Initial Loss: {history.history['loss'][0]:.4f} | Final Loss (Epoch 10): {history.history['loss'][-1]:.4f}")
```

```python
# =====================================================================
# Question 23: Reinforcement learning agent for traffic lights
# =====================================================================
import numpy as np

print("=== QUESTION 23: Q-learning Reinforcement Learning Agent ===")
states = [0, 1]   # 0: Low Traffic, 1: High Traffic
actions = [0, 1]  # 0: Keep Green Duration, 1: Increase Duration
Q = np.zeros((len(states), len(actions)))

learning_rate = 0.1
discount_factor = 0.9
state = 0

for episode in range(100):
    action = np.argmax(Q[state]) if np.random.rand() > 0.2 else np.random.choice(actions)
    reward = 10 if (state == 1 and action == 1) or (state == 0 and action == 0) else -2
    next_state = np.random.choice(states)

    # Bellman Optimality Equation Update
    Q[state, action] += learning_rate * (reward + discount_factor * np.max(Q[next_state]) - Q[state, action])
    state = next_state

print("Simulated Traffic Q-Table Result:\n", Q)
```

```python
# =====================================================================
# Question 25: Solve XOR using Multilayer Feedforward Network
# =====================================================================
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

print("=== QUESTION 25: Solving XOR logic with Feedforward Network ===")
X_xor = np.array([[0,0], [0,1], [1,0], [1,1]])
y_xor = np.array([[0], [1], [1], [0]])

xor_model = Sequential([
    Input(shape=(2,)),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])
xor_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
xor_model.fit(X_xor, y_xor, epochs=500, verbose=0)

print("Predictions for XOR:")
for val in X_xor:
    pred = xor_model.predict(np.array([val]), verbose=0)[0][0]
    print(f"{val} -> {int(np.round(pred))} (Confidence: {pred:.4f})")
```

```python
# =====================================================================
# Question 26: Classification on Iris Dataset
# =====================================================================
from sklearn.datasets import load_iris
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

print("=== QUESTION 26: Iris Dataset Neural Network Classification ===")
iris = load_iris()
X_iris, y_iris = iris.data, iris.target.reshape(-1, 1)
y_iris_enc = OneHotEncoder(sparse_output=False).fit_transform(y_iris)

X_train_ir, X_test_ir, y_train_ir, y_test_ir = train_test_split(X_iris, y_iris_enc, test_size=0.2, random_state=42)

iris_model = Sequential([
    Input(shape=(4,)),
    Dense(16, activation='relu'),
    Dense(8, activation='relu'),
    Dense(3, activation='softmax')
])
iris_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
iris_model.fit(X_train_ir, y_train_ir, epochs=50, batch_size=4, verbose=0)
loss_ir, acc_ir = iris_model.evaluate(X_test_ir, y_test_ir, verbose=0)
print(f"Iris Dataset Multiclass Accuracy: {acc_ir * 100:.2f}%")
```

```python
# =====================================================================
# Question 27: Implement and visualize activation functions
# =====================================================================
import numpy as np
import matplotlib.pyplot as plt

print("=== QUESTION 27: Visualizing Activation Functions ===")
x_vals = np.linspace(-5, 5, 100)
sigmoid_vals = 1 / (1 + np.exp(-x_vals))
tanh_vals = np.tanh(x_vals)
relu_vals = np.maximum(0, x_vals)

plt.figure(figsize=(8, 4))
plt.plot(x_vals, sigmoid_vals, label='Sigmoid')
plt.plot(x_vals, tanh_vals, label='Tanh')
plt.plot(x_vals, relu_vals, label='ReLU')
plt.title("Activation Curve Distributions")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# =====================================================================
# Question 29: Handcrafted forward propagation
# =====================================================================
import numpy as np

print("=== QUESTION 29: Manual Forward Propagation calculation ===")
X_in = np.array([0.5, 0.8])
W1 = np.array([[0.2, 0.4], [0.6, 0.1]])
b1 = np.array([0.1, -0.2])
W2 = np.array([0.5, -0.3])
b2 = 0.05

# Hidden representation layers
hidden_in = np.dot(X_in, W1) + b1
hidden_out = 1 / (1 + np.exp(-hidden_in)) # Sigmoid activation

# Output activation step
output_in = np.dot(hidden_out, W2) + b2
output_out = 1 / (1 + np.exp(-output_in))

print("Input vector:", X_in)
print("Hidden layer output matrix:", hidden_out)
print("Output layer activation:", output_out)
```

```python
# =====================================================================
# Question 30: Backpropagation implementation details
# =====================================================================
import numpy as np

print("=== QUESTION 30: Manual Backpropagation Algorithm Step ===")
x_val = np.array([1.0, 2.0])
y_true = np.array([0.5])

w_hidden = np.random.randn(2, 2)
w_output = np.random.randn(2, 1)

# Forward
h_out = 1 / (1 + np.exp(-np.dot(x_val, w_hidden)))
o_out = 1 / (1 + np.exp(-np.dot(h_out, w_output)))

# Backprop local delta
loss_err = o_out - y_true
d_output = loss_err * (o_out * (1 - o_out))

w_output_grad = np.dot(h_out.reshape(-1, 1), d_output.reshape(1, -1))
print("Computed Hidden-to-Output Weight Gradient Matrices:\n", w_output_grad)
```
