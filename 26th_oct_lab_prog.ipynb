```python
import numpy as np
import matplotlib.pyplot as plt

# Define the covariance matrix C
C = np.array([
    [5, 2],
    [2, 2]
])

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)

# Find index of the largest eigenvalue
largest_idx = np.argmax(eigenvalues)
largest_eigenvalue = eigenvalues[largest_idx]
principal_eigenvector = eigenvectors[:, largest_idx]

# Display computed results
print("Covariance Matrix C:")
print(C)
print("\nEigenvalues:", eigenvalues)
print("\nEigenvectors (as columns):\n", eigenvectors)
print(f"\nLargest Eigenvalue: {largest_eigenvalue:.4f}")
print(f"Principal Eigenvector (Direction of Maximum Variation): {principal_eigenvector}")

# Visualize eigenvectors
plt.figure(figsize=(7, 7))
origin = np.array([[0, 0], [0, 0]])  # Origin for quiver plotting

# Plot both eigenvectors
plt.quiver(*origin, eigenvectors[0, :], eigenvectors[1, :],
           color=['gray', 'gray'], scale=1, scale_units='xy', angles='xy',
           alpha=0.5, label='Eigenvectors')

# Highlight the principal eigenvector
plt.quiver(0, 0, principal_eigenvector[0], principal_eigenvector[1],
           color='green', scale=1, scale_units='xy', angles='xy',
           label=f'Principal Eigenvector (\u03bb = {largest_eigenvalue:.1f})')

# Final graph setups
max_val = np.max(np.abs(eigenvectors)) * 1.5
plt.xlim(-max_val, max_val)
plt.ylim(-max_val, max_val)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.grid(True, linestyle=':', alpha=0.6)
plt.gca().set_aspect('equal')
plt.xlabel('Sensor 1 dimension')
plt.ylabel('Sensor 2 dimension')
plt.title('Principal Direction & Eigenvectors of Sensor Covariance Matrix')
plt.legend()
plt.show()
```

### Lab Question 11 (Probability Distributions and Bayes' Theorem - Q1)

**Problem Statement:**
A factory produces electronic components, and the probability that a component is defective is $0.1$. A quality inspector randomly selects $20$ components. The number of defective components follows a binomial distribution.

Write a Python program to calculate the probability that exactly $3$ components are defective, and to plot the probability mass function for $0$ to $20$ defective components.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parameters of the binomial distribution
n = 20  # total number of trials (components)
p = 0.1 # probability of success (defective component)
k = 3   # number of interest

# 1. Calculate the probability of exactly k = 3 defective components
prob_exact_3 = binom.pmf(k, n, p)
print(f"Probability of exactly {k} defective components out of {n}: {prob_exact_3:.6f} (or {prob_exact_3*100:.4f}%)")

# 2. Generate values from 0 to 20 for plotting the PMF
x_values = np.arange(0, n + 1)
pmf_values = binom.pmf(x_values, n, p)

# Plot the PMF
plt.figure(figsize=(10, 6))
plt.stem(x_values, pmf_values, linefmt='b-', markerfmt='bo', basefmt='r-')
plt.title('Binomial Distribution PMF (n=20, p=0.1)')
plt.xlabel('Number of Defective Components')
plt.ylabel('Probability')
plt.xticks(x_values)
plt.grid(True, linestyle='--', alpha=0.5)

# Highlight k=3
plt.axvline(x=3, color='orange', linestyle='--', linewidth=1.5, label='k = 3')
plt.legend()
plt.show()
```

### Lab Question 12 (Probability Distributions and Bayes' Theorem - Q2)

**Problem Statement:**
A lecturer observes that students’ examination marks approximately follow a normal distribution with a mean of $70$ and a standard deviation of $10$.

Write a Python program to calculate the probability that a randomly selected student scores between $60$ and $80$, and to plot the normal distribution with the interval highlighted.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Distribution parameters
mean = 70
std_dev = 10

# Define the interval boundaries
lower_bound = 60
upper_bound = 80

# 1. Calculate cumulative probabilities
prob_lower = norm.cdf(lower_bound, loc=mean, scale=std_dev)
prob_upper = norm.cdf(upper_bound, loc=mean, scale=std_dev)

# Probability of scoring between 60 and 80
prob_interval = prob_upper - prob_lower
print(f"Probability of a student scoring between {lower_bound} and {upper_bound}: {prob_interval:.6f} (or {prob_interval*100:.4f}%)")

# 2. Define data points for plotting the PDF curve
x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)
y = norm.pdf(x, loc=mean, scale=std_dev)

# Plot the Normal Distribution PDF
plt.figure(figsize=(10, 6))
plt.plot(x, y, color='blue', linewidth=2, label=f'Normal Distribution (\u03bc={mean}, \u03c3={std_dev})')

# Highlight the interval [60, 80]
x_fill = np.linspace(lower_bound, upper_bound, 500)
y_fill = norm.pdf(x_fill, loc=mean, scale=std_dev)
plt.fill_between(x_fill, y_fill, color='green', alpha=0.3, label=f'Interval {lower_bound} to {upper_bound}')

# Add lines for boundaries and mean
plt.axvline(x=mean, color='black', linestyle='--', linewidth=1.5, label=f'Mean (\u03bc = {mean})')
plt.axvline(x=lower_bound, color='red', linestyle=':', linewidth=1.5)
plt.axvline(x=upper_bound, color='red', linestyle=':', linewidth=1.5)

# Set title and labels
plt.title('Normal Distribution of Exam Marks with Highlighted Interval')
plt.xlabel('Examination Marks')
plt.ylabel('Probability Density')
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 14 (Probability Distributions and Bayes' Theorem - Q4)

**Problem Statement:**
A disease affects $2\%$ of a population ($P(\text{Disease}) = 0.02$). A medical test correctly identifies a person with the disease $95\%$ of the time ($P(\text{Positive} \mid \text{Disease}) = 0.95$). The test also produces a positive result for $5\%$ of people who do not have the disease ($P(\text{Positive} \mid \text{No Disease}) = 0.05$).

Write a Python program to apply Bayes’ theorem and calculate the probability that a person actually has the disease given that the test result is positive ($P(\text{Disease} \mid \text{Positive})$). Use Matplotlib to compare the prior probability of having the disease with the posterior probability after a positive test result.

```python
import matplotlib.pyplot as plt

# 1. Given values (priors and conditionals)
prior_disease = 0.02
prior_no_disease = 1.0 - prior_disease

sensitivity = 0.95       # P(Positive | Disease)
false_positive_rate = 0.05 # P(Positive | No Disease)

# 2. Apply Bayes' Theorem
# Total probability of testing positive: P(Positive) = P(Positive|Disease)*P(Disease) + P(Positive|No Disease)*P(No Disease)
total_positive = (sensitivity * prior_disease) + (false_positive_rate * prior_no_disease)

# Posterior probability: P(Disease | Positive)
posterior_disease = (sensitivity * prior_disease) / total_positive

print(f"Prior Probability of having the disease: {prior_disease:.4f} ({prior_disease*100:.1f}%)")
print(f"Total Probability of testing positive: {total_positive:.4f} ({total_positive*100:.2f}%)")
print(f"Posterior Probability of having disease given Positive test: {posterior_disease:.6f} ({posterior_disease*100:.4f}%)")

# 3. Visualize using Matplotlib
probabilities = [prior_disease, posterior_disease]
labels = ['Prior Probability\nP(Disease)', 'Posterior Probability\nP(Disease | Positive)']

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, probabilities, color=['skyblue', 'lightcoral'], width=0.4)

# Add values on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.02, f'{height:.4f} ({height*100:.2f}%)',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.title("Comparison of Prior and Posterior Probabilities (Disease Medical Test)", fontsize=14, pad=15)
plt.ylabel("Probability Value", fontsize=12)
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 15 (Probability Distributions and Bayes' Theorem - Q5)

**Problem Statement:**
An email filtering system identifies $30\%$ of incoming emails as spam ($P(\text{Spam}) = 0.30$). The word “prize” appears in $70\%$ of spam emails ($P(\text{"prize"} \mid \text{Spam}) = 0.70$) and in $10\%$ of non-spam emails ($P(\text{"prize"} \mid \text{Ham}) = 0.10$).

Write a Python program to apply Bayes’ theorem and calculate the probability that an email is spam given that it contains the word “prize” ($P(\text{Spam} \mid \text{"prize"})$). Use Matplotlib to visualize the prior probability of spam and the posterior probability after observing the word.

```python
import matplotlib.pyplot as plt

# 1. Given values (priors and conditionals)
prior_spam = 0.30
prior_ham = 1.0 - prior_spam

prob_prize_given_spam = 0.70  # P("prize" | Spam)
prob_prize_given_ham = 0.10   # P("prize" | Ham)

# 2. Apply Bayes' Theorem
# Total probability of detecting "prize": P("prize") = P("prize"|Spam)*P(Spam) + P("prize"|Ham)*P(Ham)
total_prize = (prob_prize_given_spam * prior_spam) + (prob_prize_given_ham * prior_ham)

# Posterior probability: P(Spam | "prize")
posterior_spam = (prob_prize_given_spam * prior_spam) / total_prize

print(f"Prior Probability of Spam: {prior_spam:.4f} ({prior_spam*100:.1f}%)")
print(f"Total Probability of observing 'prize': {total_prize:.4f} ({total_prize*100:.1f}%)")
print(f"Posterior Probability of Spam given 'prize' exists: {posterior_spam:.6f} ({posterior_spam*100:.4f}%)")

# 3. Visualize using Matplotlib
probabilities = [prior_spam, posterior_spam]
labels = ['Prior Probability\nP(Spam)', 'Posterior Probability\nP(Spam | "prize")']

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, probabilities, color=['lightgray', 'orange'], width=0.4)

# Add values on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.02, f'{height:.4f} ({height*100:.2f}%)',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.title("Comparison of Prior and Posterior Probabilities (Spam Filter)", fontsize=14, pad=15)
plt.ylabel("Probability Value", fontsize=12)
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 17 (Entropy, Cross-Entropy and KL Divergence - Q2)

**Problem Statement:**
To compare model complexity and predictions, a machine learning model predicts the weather probabilities as $Q=[0.4, 0.4, 0.2]$, while the actual probabilities are $P=[0.6, 0.3, 0.1]$.

Write a Python program to calculate the cross-entropy $H(P, Q)$ between the true distribution $P$ and predicted distribution $Q$ to measure the prediction error. Use Matplotlib to plot a comparison bar chart of both distributions.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define true distribution P and predicted distribution Q
P = np.array([0.6, 0.3, 0.1])
Q = np.array([0.4, 0.4, 0.2])
categories = ['Sunny', 'Rainy', 'Cloudy']

# 2. Calculate Cross-Entropy H(P, Q) in bits
# H(P, Q) = -sum(p_i * log2(q_i))
cross_entropy = -np.sum(P * np.log2(Q))

print(f"True Distribution P: {P}")
print(f"Predicted Distribution Q: {Q}")
print(f"Calculated Cross-Entropy H(P, Q): {cross_entropy:.6f} bits")

# 3. Plot grouped bar chart comparing P and Q
x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(10, 6))
plt.bar(x - width/2, P, width, label='True Distribution (P)', color='royalblue')
plt.bar(x + width/2, Q, width, label='Predicted Distribution (Q)', color='salmon')

plt.title(f'Comparison of True (P) vs Predicted (Q) Weather Probabilities\nCross-Entropy H(P, Q) = {cross_entropy:.4f} bits', fontsize=13)
plt.xlabel('Weather Condition')
plt.ylabel('Probability')
plt.xticks(x, categories)
plt.ylim(0, 1.0)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 18 (Entropy, Cross-Entropy and KL Divergence - Q3)

**Problem Statement:**
A data scientist compares a true probability distribution $P=[0.7, 0.2, 0.1]$ of a dataset with an approximated distribution $Q=[0.5, 0.3, 0.2]$.

Write a Python program to calculate the Kullback-Leibler (KL) divergence $D_{KL}(P \parallel Q)$ to measure the information lost when approximating $P$ with $Q$. Use Matplotlib to plot the KL divergence contribution of each category.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define distributions
P = np.array([0.7, 0.2, 0.1])
Q = np.array([0.5, 0.3, 0.2])
categories = ['Class A', 'Class B', 'Class C']

# 2. Calculate KL Divergence contributions: p_i * log2(p_i / q_i)
kl_contributions = P * np.log2(P / Q)
kl_divergence = np.sum(kl_contributions)

print(f"True Distribution P: {P}")
print(f"Approximate Distribution Q: {Q}")
print(f"KL Contributions per category: {kl_contributions}")
print(f"Total KL Divergence D_KL(P || Q): {kl_divergence:.6f} bits")

# 3. Plot the contributions
plt.figure(figsize=(8, 5))
plt.bar(categories, kl_contributions, color='purple', width=0.4, edgecolor='black')
plt.axhline(0, color='black', linewidth=1)
plt.title(f'KL Divergence Contributions per Category\nTotal $D_{{KL}}(P \\parallel Q)$ = {kl_divergence:.4f} bits', fontsize=13)
plt.xlabel('Categories')
plt.ylabel('Contribution (bits)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 22 (Entropy, Cross-Entropy and KL Divergence - Q5)

**Problem Statement:**
A neural network classifies an image into four categories. The actual class distribution is $Y=[0,1,0,0]$, and the predicted probability distribution is $P=[0.1,0.6,0.2,0.1]$.

Write a Python program to calculate the cross-entropy loss and KL divergence between the actual and predicted distributions, and to visualize both distributions.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define actual and predicted distributions
Y = np.array([0, 1, 0, 0])
P = np.array([0.1, 0.6, 0.2, 0.1])
categories = ['Class A', 'Class B', 'Class C', 'Class D']

# 2. Calculations (using natural log for standard cross-entropy and KL divergence)
cross_entropy_loss = -np.sum(Y * np.log(P))

# KL Divergence: sum(Y_i * ln(Y_i / P_i))
kl_div = 0.0
for y_val, p_val in zip(Y, P):
    if y_val > 0:
        kl_div += y_val * np.log(y_val / p_val)

print(f"Actual Distribution (Y): {Y}")
print(f"Predicted Distribution (P): {P}")
print(f"Cross-Entropy Loss: {cross_entropy_loss:.6f} nats")
print(f"KL Divergence D_KL(Y || P): {kl_div:.6f} nats")

# 3. Visualization
x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(9, 5))
plt.bar(x - width/2, Y, width, label='Actual (Y)', color='mediumseagreen')
plt.bar(x + width/2, P, width, label='Predicted (P)', color='tomato')

plt.title(f'Actual vs Predicted Distribution\nCE Loss = {cross_entropy_loss:.4f} nats | $D_{{KL}}(Y \\parallel P)$ = {kl_div:.4f} nats')
plt.xlabel('Classes')
plt.ylabel('Probability')
plt.xticks(x, categories)
plt.ylim(0, 1.1)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 25 (Bias-Variance Section - Q3)

**Problem Statement:**
Write a Python program to generate a sample nonlinear dataset, train polynomial regression models with degrees 2 and 15, calculate their training and testing errors, and compare their fitted curves.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# 1. Generate synthetic non-linear dataset
np.random.seed(10)
n_samples = 35
X = np.sort(np.random.rand(n_samples) * 6)
y = np.sin(X) + np.random.randn(n_samples) * 0.25

X_reshaped = X.reshape(-1, 1)

# Split into Train and Test
X_train, X_test, y_train, y_test = train_test_split(X_reshaped, y, test_size=0.3, random_state=42)

# Range for plotting curves
X_plot = np.linspace(0, 6, 200).reshape(-1, 1)

# 2. Train and evaluate polynomial models
degrees = [2, 15]
plt.figure(figsize=(12, 6))
plt.scatter(X_train, y_train, color='blue', label='Train Data')
plt.scatter(X_test, y_test, color='green', marker='s', label='Test Data')

for deg in degrees:
    model = make_pipeline(PolynomialFeatures(degree=deg), LinearRegression())
    model.fit(X_train, y_train)

    # Calculations
    train_mse = mean_squared_error(y_train, model.predict(X_train))
    test_mse = mean_squared_error(y_test, model.predict(X_test))

    print(f"Degree {deg:2d} -> Train MSE: {train_mse:.4f} | Test MSE: {test_mse:.4f}")

    # Plot curves
    y_plot = model.predict(X_plot)
    plt.plot(X_plot, y_plot, label=f'Degree {deg} (Train MSE: {train_mse:.2f}, Test MSE: {test_mse:.2f})', linewidth=2)

plt.title('Comparison of Underfitting (Degree 2) vs Overfitting (Degree 15)', fontsize=14)
plt.xlabel('Feature X')
plt.ylabel('Target y')
plt.ylim(-2, 2)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 27 (Bias-Variance Section - Q5)

**Problem Statement:**
Write a Python program to generate a sample regression dataset, calculate learning curves for a linear regression model (low complexity) and a high-degree polynomial regression model (high complexity), and plot training and validation errors against the number of training samples.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import learning_curve

# 1. Generate synthetic regression dataset
np.random.seed(42)
n_samples = 150
X = np.sort(np.random.rand(n_samples) * 10)
y = 5 * (X**2) - 2 * X + np.random.randn(n_samples) * 15

X_reshaped = X.reshape(-1, 1)

# Define the learning curves calculator function
def plot_model_learning_curve(model, name, subplot_idx):
    train_sizes, train_scores, validation_scores = learning_curve(
        model, X_reshaped, y, train_sizes=np.linspace(0.1, 1.0, 10),
        cv=5, scoring='neg_mean_squared_error'
    )

    # Convert negative MSE scores back to positive MSE
    train_errors = -train_scores.mean(axis=1)
    val_errors = -validation_scores.mean(axis=1)

    plt.subplot(1, 2, subplot_idx)
    plt.plot(train_sizes, train_errors, 'r-o', label='Training error')
    plt.plot(train_sizes, val_errors, 'b-s', label='Validation error')
    plt.title(f'Learning Curve: {name}')
    plt.xlabel('Training Set Size')
    plt.ylabel('Mean Squared Error')
    plt.legend()
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.ylim(0, 800)  # Bound y-axis to compare on same scale

# 2. Run learning curves calculations and plot side-by-side
plt.figure(figsize=(15, 6))

# Linear Regression Model
plot_model_learning_curve(LinearRegression(), 'Linear Model (High Bias / Underfitting)', 1)

# Degree 5 Polynomial Regression Model
poly_model = make_pipeline(PolynomialFeatures(degree=5), LinearRegression())
plot_model_learning_curve(poly_model, 'Degree 5 Polynomial Model (Balanced/Optimal)', 2)

plt.tight_layout()
plt.show()
```

Write a Python program to calculate the cross-entropy loss and KL divergence between the actual and predicted distributions, and to visualize both distributions.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define actual and predicted distributions
Y = np.array([0, 1, 0, 0])
P = np.array([0.1, 0.6, 0.2, 0.1])
categories = ['Class A', 'Class B', 'Class C', 'Class D']

# 2. Calculations (using natural log for standard cross-entropy and KL divergence)
# We handle potential division/log of 0 safely where needed
# Cross-Entropy Loss: -sum(Y_i * ln(P_i))
cross_entropy_loss = -np.sum(Y * np.log(P))

# KL Divergence: sum(Y_i * ln(Y_i / P_i))
# Since Y has zeros, we only sum where Y_i > 0
kl_div = 0.0
for y_val, p_val in zip(Y, P):
    if y_val > 0:
        kl_div += y_val * np.log(y_val / p_val)

print(f"Actual Distribution (Y): {Y}")
print(f"Predicted Distribution (P): {P}")
print(f"Cross-Entropy Loss: {cross_entropy_loss:.6f} nats")
print(f"KL Divergence D_KL(Y || P): {kl_div:.6f} nats")

# 3. Visualization
x = np.arange(len(categories))
width = 0.35

plt.figure(figsize=(9, 5))
plt.bar(x - width/2, Y, width, label='Actual (Y)', color='mediumseagreen')
plt.bar(x + width/2, P, width, label='Predicted (P)', color='tomato')

plt.title(f'Actual vs Predicted Distribution\nCE Loss = {cross_entropy_loss:.4f} nats | $D_{{KL}}(Y \\parallel P)$ = {kl_div:.4f} nats')
plt.xlabel('Classes')
plt.ylabel('Probability')
plt.xticks(x, categories)
plt.ylim(0, 1.1)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### Lab Question 16 (Entropy, Cross-Entropy and KL Divergence - Q1)

**Problem Statement:**
A weather prediction system estimates the probabilities of three possible conditions as sunny, rainy, and cloudy: $P=[0.6, 0.3, 0.1]$. The system needs to measure the uncertainty in its predictions.

Write a Python program to calculate the entropy of the probability distribution and to display the probabilities of the three weather conditions in a bar chart.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the probability distribution P
P = np.array([0.6, 0.3, 0.1])
conditions = ['Sunny', 'Rainy', 'Cloudy']

# 2. Calculate the Shannon Entropy in bits (log base 2)
# Entropy H(P) = -sum(p_i * log2(p_i))
# We filter out 0 probabilities if they exist, though not present here
entropy = -np.sum(P * np.log2(P))

print(f"Probability Distribution P: {dict(zip(conditions, P))}")
print(f"Calculated Shannon Entropy: {entropy:.6f} bits")

# 3. Create a bar chart showing the probabilities
plt.figure(figsize=(8, 5))
bars = plt.bar(conditions, P, color=['gold', 'lightskyblue', 'lightgray'], width=0.4, edgecolor='gray')

# Add values on top of the bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.015, f'{height:.1f} ({height*100:.0f}%)',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.title(f"Weather Condition Probabilities (Entropy = {entropy:.4f} bits)", fontsize=13, pad=15)
plt.xlabel("Weather Condition", fontsize=12)
plt.ylabel("Probability", fontsize=12)
plt.ylim(0, 1.0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
